# 4. Conversión a aprendizaje supervisado

Una LSTM recibe secuencias. Con una ventana \(w\), cada ejemplo utiliza los últimos \(w\) meses para estimar el mes siguiente:

\[
\left(y_{t-w},\ldots,y_{t-1}\right)\longrightarrow y_t.
\]

Se prueban ventanas de 12 y 24 meses. La primera permite observar un ciclo anual completo; la segunda ofrece dos ciclos y más contexto, aunque reduce el número de ejemplos disponibles.


```python
def crear_secuencias(valores_escalados, fechas, ventana):
    X, y, fechas_objetivo, posiciones = [], [], [], []
    for t in range(ventana, len(valores_escalados)):
        X.append(valores_escalados[t - ventana:t])
        y.append(valores_escalados[t])
        fechas_objetivo.append(fechas[t])
        posiciones.append(t)
    return (
        np.asarray(X, dtype=float),
        np.asarray(y, dtype=float),
        pd.DatetimeIndex(fechas_objetivo),
        np.asarray(posiciones),
    )


def preparar_datos_serie(serie, ventana):
    train_original = serie.iloc[:N_TRAIN_ORIGINAL].astype(float)
    scaler = MinMaxScaler(feature_range=(-1, 1))
    scaler.fit(train_original.iloc[:N_CORE].to_numpy().reshape(-1, 1))

    escalados = scaler.transform(
        train_original.to_numpy().reshape(-1, 1)
    ).ravel()
    X, y, fechas_objetivo, posiciones = crear_secuencias(
        escalados, train_original.index, ventana
    )
    mascara_train = posiciones < N_CORE
    mascara_val = posiciones >= N_CORE
    return {
        "X_train": X[mascara_train],
        "y_train": y[mascara_train],
        "X_val": X[mascara_val],
        "y_val": y[mascara_val],
        "fechas_val": fechas_objetivo[mascara_val],
        "scaler": scaler,
    }


ejemplo_12 = preparar_datos_serie(serie_total, ventana=12)
ejemplo_24 = preparar_datos_serie(serie_total, ventana=24)
tabla_ventanas = pd.DataFrame({
    "Ventana": [12, 24],
    "Ejemplos de ajuste": [
        len(ejemplo_12["X_train"]), len(ejemplo_24["X_train"])
    ],
    "Ejemplos de validación": [
        len(ejemplo_12["X_val"]), len(ejemplo_24["X_val"])
    ],
    "Forma de entrada": [
        str(ejemplo_12["X_train"].shape),
        str(ejemplo_24["X_train"].shape),
    ],
})
display(tabla_ventanas)
```

# 5. Implementación reproducible de la LSTM

En cada instante \(t\), la celda calcula:

\[
\begin{aligned}
f_t &= \sigma(W_f[h_{t-1},x_t]+b_f),\\
i_t &= \sigma(W_i[h_{t-1},x_t]+b_i),\\
\tilde c_t &= \tanh(W_g[h_{t-1},x_t]+b_g),\\
c_t &= f_t\odot c_{t-1}+i_t\odot \tilde c_t,\\
o_t &= \sigma(W_o[h_{t-1},x_t]+b_o),\\
h_t &= o_t\odot\tanh(c_t).
\end{aligned}
\]

La implementación admite una o dos capas LSTM, inicialización reproducible, regularización L2, recorte global de gradiente, optimizador Adam y parada temprana. La función de pérdida utilizada para entrenar es el error cuadrático medio en la escala normalizada.


```python
def sigmoid(x):
    x = np.clip(x, -40, 40)
    return 1.0 / (1.0 + np.exp(-x))


class LSTMNumPyRegressor:
    '''Regresor secuencia-a-un-valor con una o varias capas LSTM.'''

    def __init__(
        self,
        hidden_sizes=(12,),
        learning_rate=0.005,
        l2=1e-4,
        batch_size=16,
        max_epochs=90,
        patience=12,
        clip_norm=5.0,
        seed=2026,
    ):
        self.hidden_sizes = tuple(hidden_sizes)
        self.learning_rate = float(learning_rate)
        self.l2 = float(l2)
        self.batch_size = int(batch_size)
        self.max_epochs = int(max_epochs)
        self.patience = int(patience)
        self.clip_norm = float(clip_norm)
        self.seed = int(seed)
        self.rng = np.random.default_rng(seed)

    def _initialize(self):
        self.layers = []
        input_size = 1
        for hidden_size in self.hidden_sizes:
            scale = np.sqrt(2.0 / (input_size + hidden_size))
            W = self.rng.normal(
                0, scale, size=(4 * hidden_size, hidden_size + input_size)
            )
            b = np.zeros(4 * hidden_size)
            b[:hidden_size] = 1.0  # sesgo inicial de la puerta de olvido
            self.layers.append({"W": W, "b": b})
            input_size = hidden_size

        out_scale = np.sqrt(2.0 / max(1, self.hidden_sizes[-1]))
        self.Wy = self.rng.normal(0, out_scale, size=(1, self.hidden_sizes[-1]))
        self.by = np.zeros(1)

        self._adam_m = {
            "layers": [
                {"W": np.zeros_like(p["W"]), "b": np.zeros_like(p["b"])}
                for p in self.layers
            ],
            "Wy": np.zeros_like(self.Wy),
            "by": np.zeros_like(self.by),
        }
        self._adam_v = copy.deepcopy(self._adam_m)
        self._adam_t = 0

    def _forward_one(self, sequence):
        states = []
        for h in self.hidden_sizes:
            states.append({"h": np.zeros(h), "c": np.zeros(h)})

        caches = [[] for _ in self.layers]
        for x_t in np.asarray(sequence).ravel():
            inp = np.array([x_t], dtype=float)
            for layer_index, (params, hidden_size) in enumerate(
                zip(self.layers, self.hidden_sizes)
            ):
                h_prev = states[layer_index]["h"]
                c_prev = states[layer_index]["c"]
                z = np.concatenate([h_prev, inp])
                a = params["W"] @ z + params["b"]
                af, ai, ao, ag = np.split(a, 4)
                f = sigmoid(af)
                i = sigmoid(ai)
                o = sigmoid(ao)
                g = np.tanh(ag)
                c = f * c_prev + i * g
                h = o * np.tanh(c)
                caches[layer_index].append({
                    "z": z, "h_prev": h_prev, "c_prev": c_prev,
                    "f": f, "i": i, "o": o, "g": g, "c": c, "h": h,
                })
                states[layer_index] = {"h": h, "c": c}
                inp = h

        h_top = states[-1]["h"]
        y_hat = float((self.Wy @ h_top + self.by)[0])
        return y_hat, caches, h_top

    def _zero_grads(self):
        return {
            "layers": [
                {"W": np.zeros_like(p["W"]), "b": np.zeros_like(p["b"])}
                for p in self.layers
            ],
            "Wy": np.zeros_like(self.Wy),
            "by": np.zeros_like(self.by),
        }

    def _backward_one(self, y_hat, y_true, caches, h_top):
        grads = self._zero_grads()
        dy = y_hat - float(y_true)
        grads["Wy"] += dy * h_top.reshape(1, -1)
        grads["by"] += dy

        dh_future = [np.zeros(h) for h in self.hidden_sizes]
        dc_future = [np.zeros(h) for h in self.hidden_sizes]
        top = len(self.layers) - 1
        T = len(caches[0])

        for t in range(T - 1, -1, -1):
            gradient_from_upper = None
            for layer_index in range(top, -1, -1):
                cache = caches[layer_index][t]
                h_size = self.hidden_sizes[layer_index]
                dh = dh_future[layer_index].copy()
                if layer_index == top and t == T - 1:
                    dh += self.Wy.ravel() * dy
                if gradient_from_upper is not None:
                    dh += gradient_from_upper

                tanh_c = np.tanh(cache["c"])
                do = dh * tanh_c
                dc = dh * cache["o"] * (1.0 - tanh_c**2) + dc_future[layer_index]
                df = dc * cache["c_prev"]
                di = dc * cache["g"]
                dg = dc * cache["i"]

                daf = df * cache["f"] * (1.0 - cache["f"])
                dai = di * cache["i"] * (1.0 - cache["i"])
                dao = do * cache["o"] * (1.0 - cache["o"])
                dag = dg * (1.0 - cache["g"]**2)
                da = np.concatenate([daf, dai, dao, dag])

                grads["layers"][layer_index]["W"] += np.outer(da, cache["z"])
                grads["layers"][layer_index]["b"] += da

                dz = self.layers[layer_index]["W"].T @ da
                dh_future[layer_index] = dz[:h_size]
                dc_future[layer_index] = dc * cache["f"]
                gradient_from_upper = dz[h_size:]

        return grads, 0.5 * dy**2

    @staticmethod
    def _add_grads(target, source):
        for t, s in zip(target["layers"], source["layers"]):
            t["W"] += s["W"]
            t["b"] += s["b"]
        target["Wy"] += source["Wy"]
        target["by"] += source["by"]

    def _regularize_and_scale(self, grads, batch_n):
        for grad, param in zip(grads["layers"], self.layers):
            grad["W"] = grad["W"] / batch_n + self.l2 * param["W"]
            grad["b"] /= batch_n
        grads["Wy"] = grads["Wy"] / batch_n + self.l2 * self.Wy
        grads["by"] /= batch_n

        arrays = [grads["Wy"], grads["by"]]
        for grad in grads["layers"]:
            arrays.extend([grad["W"], grad["b"]])
        norm = np.sqrt(sum(float(np.sum(a * a)) for a in arrays))
        if norm > self.clip_norm:
            factor = self.clip_norm / (norm + 1e-12)
            for array in arrays:
                array *= factor

    def _adam_update(self, grads):
        self._adam_t += 1
        beta1, beta2, eps = 0.9, 0.999, 1e-8

        triplets = []
        for idx in range(len(self.layers)):
            for key in ["W", "b"]:
                triplets.append((
                    self.layers[idx][key],
                    grads["layers"][idx][key],
                    self._adam_m["layers"][idx][key],
                    self._adam_v["layers"][idx][key],
                ))
        triplets.extend([
            (self.Wy, grads["Wy"], self._adam_m["Wy"], self._adam_v["Wy"]),
            (self.by, grads["by"], self._adam_m["by"], self._adam_v["by"]),
        ])

        for param, grad, m, v in triplets:
            m *= beta1
            m += (1 - beta1) * grad
            v *= beta2
            v += (1 - beta2) * (grad * grad)
            m_hat = m / (1 - beta1**self._adam_t)
            v_hat = v / (1 - beta2**self._adam_t)
            param -= self.learning_rate * m_hat / (np.sqrt(v_hat) + eps)

    def _snapshot(self):
        return {
            "layers": copy.deepcopy(self.layers),
            "Wy": self.Wy.copy(),
            "by": self.by.copy(),
        }

    def _restore(self, snapshot):
        self.layers = copy.deepcopy(snapshot["layers"])
        self.Wy = snapshot["Wy"].copy()
        self.by = snapshot["by"].copy()

    def predict(self, X):
        return np.asarray([self._forward_one(x)[0] for x in X], dtype=float)

    def fit(self, X_train, y_train, X_val, y_val):
        self._initialize()
        best_state = self._snapshot()
        best_val = np.inf
        wait = 0
        history = {"train_loss": [], "val_loss": []}

        for epoch in range(1, self.max_epochs + 1):
            order = self.rng.permutation(len(X_train))
            epoch_loss = 0.0

            for start in range(0, len(order), self.batch_size):
                batch_idx = order[start:start + self.batch_size]
                grads_batch = self._zero_grads()
                batch_loss = 0.0

                for idx in batch_idx:
                    y_hat, caches, h_top = self._forward_one(X_train[idx])
                    grads_one, loss_one = self._backward_one(
                        y_hat, y_train[idx], caches, h_top
                    )
                    self._add_grads(grads_batch, grads_one)
                    batch_loss += loss_one

                self._regularize_and_scale(grads_batch, len(batch_idx))
                self._adam_update(grads_batch)
                epoch_loss += batch_loss

            train_loss = epoch_loss / max(1, len(X_train))
            val_pred = self.predict(X_val)
            val_loss = float(np.mean((val_pred - y_val) ** 2))
            history["train_loss"].append(train_loss)
            history["val_loss"].append(val_loss)

            if val_loss < best_val - 1e-7:
                best_val = val_loss
                best_state = self._snapshot()
                self.best_epoch_ = epoch
                wait = 0
            else:
                wait += 1
                if wait >= self.patience:
                    break

        self._restore(best_state)
        self.history_ = history
        self.n_epochs_ = len(history["train_loss"])
        self.best_val_loss_ = best_val
        return self


print("Clase LSTM definida correctamente.")
```

# 6. Espacio de búsqueda de hiperparámetros

El tuneo es una búsqueda en rejilla pequeña y trazable. La cantidad de observaciones es limitada, por lo que se evita una red excesivamente grande.

| Código | Arquitectura | Ventana | Unidades | Tasa de aprendizaje | L2 | Lote |
|---|---|---:|---|---:|---:|---:|
| LSTM-1A | Una capa | 12 | 8 | 0.010 | 0.0001 | 16 |
| LSTM-1B | Una capa | 24 | 16 | 0.006 | 0.0003 | 16 |
| LSTM-2A | Dos capas | 12 | 12 → 6 | 0.006 | 0.0001 | 16 |
| LSTM-2B | Dos capas | 24 | 16 → 8 | 0.004 | 0.0003 | 16 |

Todas las configuraciones usan un máximo de 90 épocas, parada temprana con paciencia de 12 épocas y recorte de gradiente. El ganador se define por el **menor RMSE de validación** en la escala original de visitantes.

# 7. Tuneo y comparación en validación

Las métricas se calculan después de regresar las predicciones a la unidad original:

\[
\operatorname{MAE}=\frac{1}{n}\sum |y_t-\hat y_t|,
\qquad
\operatorname{RMSE}=\sqrt{\frac{1}{n}\sum(y_t-\hat y_t)^2}.
\]

También se reporta sMAPE para expresar el error de forma relativa y simétrica. La evaluación es de **un paso hacia adelante con historia observada**, coherente con el formato supervisado de las ventanas.


```python
def smape(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    denominador = np.abs(y_true) + np.abs(y_pred)
    return 100 * np.mean(
        np.where(denominador == 0, 0, 2 * np.abs(y_true - y_pred) / denominador)
    )


def evaluar_configuracion(nombre_serie, serie, config):
    datos = preparar_datos_serie(serie, config["ventana"])
    modelo = LSTMNumPyRegressor(
        hidden_sizes=config["hidden_sizes"],
        learning_rate=config["learning_rate"],
        l2=config["l2"],
        batch_size=config["batch_size"],
        max_epochs=90,
        patience=12,
        clip_norm=5.0,
        seed=SEED + sum(ord(c) for c in nombre_serie + config["modelo"]),
    )
    inicio = time.perf_counter()
    modelo.fit(
        datos["X_train"], datos["y_train"],
        datos["X_val"], datos["y_val"],
    )
    segundos = time.perf_counter() - inicio

    pred_escalada = modelo.predict(datos["X_val"])
    pred = datos["scaler"].inverse_transform(
        pred_escalada.reshape(-1, 1)
    ).ravel()
    real = datos["scaler"].inverse_transform(
        datos["y_val"].reshape(-1, 1)
    ).ravel()

    fila = {
        "Serie": nombre_serie,
        "Modelo": config["modelo"],
        "Familia": config["familia"],
        "Ventana": config["ventana"],
        "Unidades": " → ".join(map(str, config["hidden_sizes"])),
        "Learning rate": config["learning_rate"],
        "L2": config["l2"],
        "Lote": config["batch_size"],
        "Mejor época": modelo.best_epoch_,
        "Épocas ejecutadas": modelo.n_epochs_,
        "MAE": mean_absolute_error(real, pred),
        "RMSE": np.sqrt(mean_squared_error(real, pred)),
        "sMAPE (%)": smape(real, pred),
        "Segundos": segundos,
    }
    detalle = {
        "modelo": modelo,
        "datos": datos,
        "real": real,
        "pred": pred,
        "fechas": datos["fechas_val"],
        "config": config,
    }
    return fila, detalle


resultados = []
detalles = {}
for nombre_serie, serie in series.items():
    print(f"\nTuneando: {nombre_serie}")
    for config in CONFIGURACIONES:
        fila, detalle = evaluar_configuracion(nombre_serie, serie, config)
        resultados.append(fila)
        detalles[(nombre_serie, config["modelo"])] = detalle
        print(
            f"  {config['modelo']}: RMSE={fila['RMSE']:,.0f}, "
            f"MAE={fila['MAE']:,.0f}, época={fila['Mejor época']}"
        )

tabla_resultados = pd.DataFrame(resultados)
tabla_resultados["Posición"] = (
    tabla_resultados.groupby("Serie")["RMSE"]
    .rank(method="first", ascending=True)
    .astype(int)
)
tabla_resultados = tabla_resultados.sort_values(["Serie", "Posición"])

tabla_resultados_mostrar = tabla_resultados.copy()
tabla_resultados_mostrar["Learning rate"] = (
    tabla_resultados_mostrar["Learning rate"].map("{:.4f}".format)
)
tabla_resultados_mostrar["L2"] = tabla_resultados_mostrar["L2"].map("{:.4f}".format)
for columna in ["MAE", "RMSE"]:
    tabla_resultados_mostrar[columna] = (
        tabla_resultados_mostrar[columna].map("{:,.0f}".format)
    )
tabla_resultados_mostrar["sMAPE (%)"] = (
    tabla_resultados_mostrar["sMAPE (%)"].map("{:.2f}".format)
)
tabla_resultados_mostrar["Segundos"] = (
    tabla_resultados_mostrar["Segundos"].map("{:.2f}".format)
)
display(tabla_resultados_mostrar)
```
