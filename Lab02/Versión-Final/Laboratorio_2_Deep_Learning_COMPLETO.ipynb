# 4. Conversión a aprendizaje supervisado

Una LSTM recibe secuencias. Con una ventana \(w\), cada ejemplo utiliza los últimos \(w\) meses para estimar el mes siguiente:

\[
\left(y_{t-w},\ldots,y_{t-1}\right)\longrightarrow y_t.
\]

Se prueban ventanas de 12 y 24 meses. La primera permite observar un ciclo anual completo; la segunda ofrece dos ciclos y más contexto, aunque reduce el número de ejemplos disponibles.


```python
def crear_secuencias(valores_escalados, fechas, ventana):
    X, y, fechas_objetivo, posiciones = [], [], [], []
    for t in range(ventana, len(valores_escalados)):
        X.append(valores_escalados[t - ventana:t])
        y.append(valores_escalados[t])
        fechas_objetivo.append(fechas[t])
        posiciones.append(t)
    return (
        np.asarray(X, dtype=float),
        np.asarray(y, dtype=float),
        pd.DatetimeIndex(fechas_objetivo),
        np.asarray(posiciones),
    )


def preparar_datos_serie(serie, ventana):
    train_original = serie.iloc[:N_TRAIN_ORIGINAL].astype(float)
    scaler = MinMaxScaler(feature_range=(-1, 1))
    scaler.fit(train_original.iloc[:N_CORE].to_numpy().reshape(-1, 1))

    escalados = scaler.transform(
        train_original.to_numpy().reshape(-1, 1)
    ).ravel()
    X, y, fechas_objetivo, posiciones = crear_secuencias(
        escalados, train_original.index, ventana
    )
    mascara_train = posiciones < N_CORE
    mascara_val = posiciones >= N_CORE
    return {
        "X_train": X[mascara_train],
        "y_train": y[mascara_train],
        "X_val": X[mascara_val],
        "y_val": y[mascara_val],
        "fechas_val": fechas_objetivo[mascara_val],
        "scaler": scaler,
    }


ejemplo_12 = preparar_datos_serie(serie_total, ventana=12)
ejemplo_24 = preparar_datos_serie(serie_total, ventana=24)
tabla_ventanas = pd.DataFrame({
    "Ventana": [12, 24],
    "Ejemplos de ajuste": [
        len(ejemplo_12["X_train"]), len(ejemplo_24["X_train"])
    ],
    "Ejemplos de validación": [
        len(ejemplo_12["X_val"]), len(ejemplo_24["X_val"])
    ],
    "Forma de entrada": [
        str(ejemplo_12["X_train"].shape),
        str(ejemplo_24["X_train"].shape),
    ],
})
display(tabla_ventanas)
```

# 5. Implementación reproducible de la LSTM

En cada instante \(t\), la celda calcula:

\[
\begin{aligned}
f_t &= \sigma(W_f[h_{t-1},x_t]+b_f),\\
i_t &= \sigma(W_i[h_{t-1},x_t]+b_i),\\
\tilde c_t &= \tanh(W_g[h_{t-1},x_t]+b_g),\\
c_t &= f_t\odot c_{t-1}+i_t\odot \tilde c_t,\\
o_t &= \sigma(W_o[h_{t-1},x_t]+b_o),\\
h_t &= o_t\odot\tanh(c_t).
\end{aligned}
\]

La implementación admite una o dos capas LSTM, inicialización reproducible, regularización L2, recorte global de gradiente, optimizador Adam y parada temprana. La función de pérdida utilizada para entrenar es el error cuadrático medio en la escala normalizada.


```python
def sigmoid(x):
    x = np.clip(x, -40, 40)
    return 1.0 / (1.0 + np.exp(-x))


class LSTMNumPyRegressor:
    '''Regresor secuencia-a-un-valor con una o varias capas LSTM.'''

    def __init__(
        self,
        hidden_sizes=(12,),
        learning_rate=0.005,
        l2=1e-4,
        batch_size=16,
        max_epochs=90,
        patience=12,
        clip_norm=5.0,
        seed=2026,
    ):
        self.hidden_sizes = tuple(hidden_sizes)
        self.learning_rate = float(learning_rate)
        self.l2 = float(l2)
        self.batch_size = int(batch_size)
        self.max_epochs = int(max_epochs)
        self.patience = int(patience)
        self.clip_norm = float(clip_norm)
        self.seed = int(seed)
        self.rng = np.random.default_rng(seed)

    def _initialize(self, input_size=1):
        self.layers = []
        input_size = int(input_size)
        for hidden_size in self.hidden_sizes:
            scale = np.sqrt(2.0 / (input_size + hidden_size))
            W = self.rng.normal(
                0, scale, size=(4 * hidden_size, hidden_size + input_size)
            )
            b = np.zeros(4 * hidden_size)
            b[:hidden_size] = 1.0  # sesgo inicial de la puerta de olvido
            self.layers.append({"W": W, "b": b})
            input_size = hidden_size

        out_scale = np.sqrt(2.0 / max(1, self.hidden_sizes[-1]))
        self.Wy = self.rng.normal(0, out_scale, size=(1, self.hidden_sizes[-1]))
        self.by = np.zeros(1)

        self._adam_m = {
            "layers": [
                {"W": np.zeros_like(p["W"]), "b": np.zeros_like(p["b"])}
                for p in self.layers
            ],
            "Wy": np.zeros_like(self.Wy),
            "by": np.zeros_like(self.by),
        }
        self._adam_v = copy.deepcopy(self._adam_m)
        self._adam_t = 0

    def _forward_one(self, sequence):
        states = []
        for h in self.hidden_sizes:
            states.append({"h": np.zeros(h), "c": np.zeros(h)})

        caches = [[] for _ in self.layers]
        sequence = np.asarray(sequence, dtype=float)
        if sequence.ndim == 1:
            sequence = sequence[:, None]
        for x_t in sequence:
            inp = np.asarray(x_t, dtype=float).ravel()
            for layer_index, (params, hidden_size) in enumerate(
                zip(self.layers, self.hidden_sizes)
            ):
                h_prev = states[layer_index]["h"]
                c_prev = states[layer_index]["c"]
                z = np.concatenate([h_prev, inp])
                a = params["W"] @ z + params["b"]
                af, ai, ao, ag = np.split(a, 4)
                f = sigmoid(af)
                i = sigmoid(ai)
                o = sigmoid(ao)
                g = np.tanh(ag)
                c = f * c_prev + i * g
                h = o * np.tanh(c)
                caches[layer_index].append({
                    "z": z, "h_prev": h_prev, "c_prev": c_prev,
                    "f": f, "i": i, "o": o, "g": g, "c": c, "h": h,
                })
                states[layer_index] = {"h": h, "c": c}
                inp = h

        h_top = states[-1]["h"]
        y_hat = float((self.Wy @ h_top + self.by)[0])
        return y_hat, caches, h_top

    def _zero_grads(self):
        return {
            "layers": [
                {"W": np.zeros_like(p["W"]), "b": np.zeros_like(p["b"])}
                for p in self.layers
            ],
            "Wy": np.zeros_like(self.Wy),
            "by": np.zeros_like(self.by),
        }

    def _backward_one(self, y_hat, y_true, caches, h_top):
        grads = self._zero_grads()
        dy = y_hat - float(y_true)
        grads["Wy"] += dy * h_top.reshape(1, -1)
        grads["by"] += dy

        dh_future = [np.zeros(h) for h in self.hidden_sizes]
        dc_future = [np.zeros(h) for h in self.hidden_sizes]
        top = len(self.layers) - 1
        T = len(caches[0])

        for t in range(T - 1, -1, -1):
            gradient_from_upper = None
            for layer_index in range(top, -1, -1):
                cache = caches[layer_index][t]
                h_size = self.hidden_sizes[layer_index]
                dh = dh_future[layer_index].copy()
                if layer_index == top and t == T - 1:
                    dh += self.Wy.ravel() * dy
                if gradient_from_upper is not None:
                    dh += gradient_from_upper

                tanh_c = np.tanh(cache["c"])
                do = dh * tanh_c
                dc = dh * cache["o"] * (1.0 - tanh_c**2) + dc_future[layer_index]
                df = dc * cache["c_prev"]
                di = dc * cache["g"]
                dg = dc * cache["i"]

                daf = df * cache["f"] * (1.0 - cache["f"])
                dai = di * cache["i"] * (1.0 - cache["i"])
                dao = do * cache["o"] * (1.0 - cache["o"])
                dag = dg * (1.0 - cache["g"]**2)
                da = np.concatenate([daf, dai, dao, dag])

                grads["layers"][layer_index]["W"] += np.outer(da, cache["z"])
                grads["layers"][layer_index]["b"] += da

                dz = self.layers[layer_index]["W"].T @ da
                dh_future[layer_index] = dz[:h_size]
                dc_future[layer_index] = dc * cache["f"]
                gradient_from_upper = dz[h_size:]

        return grads, 0.5 * dy**2

    @staticmethod
    def _add_grads(target, source):
        for t, s in zip(target["layers"], source["layers"]):
            t["W"] += s["W"]
            t["b"] += s["b"]
        target["Wy"] += source["Wy"]
        target["by"] += source["by"]

    def _regularize_and_scale(self, grads, batch_n):
        for grad, param in zip(grads["layers"], self.layers):
            grad["W"] = grad["W"] / batch_n + self.l2 * param["W"]
            grad["b"] /= batch_n
        grads["Wy"] = grads["Wy"] / batch_n + self.l2 * self.Wy
        grads["by"] /= batch_n

        arrays = [grads["Wy"], grads["by"]]
        for grad in grads["layers"]:
            arrays.extend([grad["W"], grad["b"]])
        norm = np.sqrt(sum(float(np.sum(a * a)) for a in arrays))
        if norm > self.clip_norm:
            factor = self.clip_norm / (norm + 1e-12)
            for array in arrays:
                array *= factor

    def _adam_update(self, grads):
        self._adam_t += 1
        beta1, beta2, eps = 0.9, 0.999, 1e-8

        triplets = []
        for idx in range(len(self.layers)):
            for key in ["W", "b"]:
                triplets.append((
                    self.layers[idx][key],
                    grads["layers"][idx][key],
                    self._adam_m["layers"][idx][key],
                    self._adam_v["layers"][idx][key],
                ))
        triplets.extend([
            (self.Wy, grads["Wy"], self._adam_m["Wy"], self._adam_v["Wy"]),
            (self.by, grads["by"], self._adam_m["by"], self._adam_v["by"]),
        ])

        for param, grad, m, v in triplets:
            m *= beta1
            m += (1 - beta1) * grad
            v *= beta2
            v += (1 - beta2) * (grad * grad)
            m_hat = m / (1 - beta1**self._adam_t)
            v_hat = v / (1 - beta2**self._adam_t)
            param -= self.learning_rate * m_hat / (np.sqrt(v_hat) + eps)

    def _snapshot(self):
        return {
            "layers": copy.deepcopy(self.layers),
            "Wy": self.Wy.copy(),
            "by": self.by.copy(),
        }

    def _restore(self, snapshot):
        self.layers = copy.deepcopy(snapshot["layers"])
        self.Wy = snapshot["Wy"].copy()
        self.by = snapshot["by"].copy()

    def predict(self, X):
        return np.asarray([self._forward_one(x)[0] for x in X], dtype=float)

    def fit(self, X_train, y_train, X_val, y_val):
        input_size = 1 if np.asarray(X_train).ndim == 2 else X_train.shape[2]
        self._initialize(input_size=input_size)
        best_state = self._snapshot()
        best_val = np.inf
        wait = 0
        history = {"train_loss": [], "val_loss": []}

        for epoch in range(1, self.max_epochs + 1):
            order = self.rng.permutation(len(X_train))
            epoch_loss = 0.0

            for start in range(0, len(order), self.batch_size):
                batch_idx = order[start:start + self.batch_size]
                grads_batch = self._zero_grads()
                batch_loss = 0.0

                for idx in batch_idx:
                    y_hat, caches, h_top = self._forward_one(X_train[idx])
                    grads_one, loss_one = self._backward_one(
                        y_hat, y_train[idx], caches, h_top
                    )
                    self._add_grads(grads_batch, grads_one)
                    batch_loss += loss_one

                self._regularize_and_scale(grads_batch, len(batch_idx))
                self._adam_update(grads_batch)
                epoch_loss += batch_loss

            train_loss = epoch_loss / max(1, len(X_train))
            val_pred = self.predict(X_val)
            val_loss = float(np.mean((val_pred - y_val) ** 2))
            history["train_loss"].append(train_loss)
            history["val_loss"].append(val_loss)

            if val_loss < best_val - 1e-7:
                best_val = val_loss
                best_state = self._snapshot()
                self.best_epoch_ = epoch
                wait = 0
            else:
                wait += 1
                if wait >= self.patience:
                    break

        self._restore(best_state)
        self.history_ = history
        self.n_epochs_ = len(history["train_loss"])
        self.best_val_loss_ = best_val
        return self

    def fit_fixed_epochs(self, X_train, y_train, epochs):
        '''Reentrena con todos los datos durante un número fijo de épocas.'''
        input_size = 1 if np.asarray(X_train).ndim == 2 else X_train.shape[2]
        self._initialize(input_size=input_size)
        self.history_ = {"train_loss": [], "val_loss": []}

        for _ in range(int(epochs)):
            order = self.rng.permutation(len(X_train))
            epoch_loss = 0.0
            for start in range(0, len(order), self.batch_size):
                batch_idx = order[start:start + self.batch_size]
                grads_batch = self._zero_grads()
                batch_loss = 0.0
                for idx in batch_idx:
                    y_hat, caches, h_top = self._forward_one(X_train[idx])
                    grads_one, loss_one = self._backward_one(
                        y_hat, y_train[idx], caches, h_top
                    )
                    self._add_grads(grads_batch, grads_one)
                    batch_loss += loss_one
                self._regularize_and_scale(grads_batch, len(batch_idx))
                self._adam_update(grads_batch)
                epoch_loss += batch_loss
            self.history_["train_loss"].append(
                epoch_loss / max(1, len(X_train))
            )

        self.best_epoch_ = int(epochs)
        self.n_epochs_ = int(epochs)
        return self


print("Clase LSTM definida correctamente.")
```

# 7. Tuneo y comparación en validación

Las métricas se calculan después de regresar las predicciones a la unidad original:

\[
\operatorname{MAE}=\frac{1}{n}\sum |y_t-\hat y_t|,
\qquad
\operatorname{RMSE}=\sqrt{\frac{1}{n}\sum(y_t-\hat y_t)^2}.
\]

También se reporta sMAPE para expresar el error de forma relativa y simétrica. La evaluación es de **un paso hacia adelante con historia observada**, coherente con el formato supervisado de las ventanas.


```python
def smape(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    denominador = np.abs(y_true) + np.abs(y_pred)
    return 100 * np.mean(
        np.where(denominador == 0, 0, 2 * np.abs(y_true - y_pred) / denominador)
    )


def evaluar_configuracion(nombre_serie, serie, config):
    datos = preparar_datos_serie(serie, config["ventana"])
    modelo = LSTMNumPyRegressor(
        hidden_sizes=config["hidden_sizes"],
        learning_rate=config["learning_rate"],
        l2=config["l2"],
        batch_size=config["batch_size"],
        max_epochs=90,
        patience=12,
        clip_norm=5.0,
        seed=SEED + sum(ord(c) for c in nombre_serie + config["modelo"]),
    )
    inicio = time.perf_counter()
    modelo.fit(
        datos["X_train"], datos["y_train"],
        datos["X_val"], datos["y_val"],
    )
    segundos = time.perf_counter() - inicio

    pred_escalada = modelo.predict(datos["X_val"])
    pred = datos["scaler"].inverse_transform(
        pred_escalada.reshape(-1, 1)
    ).ravel()
    real = datos["scaler"].inverse_transform(
        datos["y_val"].reshape(-1, 1)
    ).ravel()

    fila = {
        "Serie": nombre_serie,
        "Modelo": config["modelo"],
        "Familia": config["familia"],
        "Ventana": config["ventana"],
        "Unidades": " → ".join(map(str, config["hidden_sizes"])),
        "Learning rate": config["learning_rate"],
        "L2": config["l2"],
        "Lote": config["batch_size"],
        "Mejor época": modelo.best_epoch_,
        "Épocas ejecutadas": modelo.n_epochs_,
        "MAE": mean_absolute_error(real, pred),
        "RMSE": np.sqrt(mean_squared_error(real, pred)),
        "sMAPE (%)": smape(real, pred),
        "Segundos": segundos,
    }
    detalle = {
        "modelo": modelo,
        "datos": datos,
        "real": real,
        "pred": pred,
        "fechas": datos["fechas_val"],
        "config": config,
    }
    return fila, detalle


resultados = []
detalles = {}
for nombre_serie, serie in series.items():
    print(f"\nTuneando: {nombre_serie}")
    for config in CONFIGURACIONES:
        fila, detalle = evaluar_configuracion(nombre_serie, serie, config)
        resultados.append(fila)
        detalles[(nombre_serie, config["modelo"])] = detalle
        print(
            f"  {config['modelo']}: RMSE={fila['RMSE']:,.0f}, "
            f"MAE={fila['MAE']:,.0f}, época={fila['Mejor época']}"
        )

tabla_resultados = pd.DataFrame(resultados)
tabla_resultados["Posición"] = (
    tabla_resultados.groupby("Serie")["RMSE"]
    .rank(method="first", ascending=True)
    .astype(int)
)
tabla_resultados = tabla_resultados.sort_values(["Serie", "Posición"])

tabla_resultados_mostrar = tabla_resultados.copy()
tabla_resultados_mostrar["Learning rate"] = (
    tabla_resultados_mostrar["Learning rate"].map("{:.4f}".format)
)
tabla_resultados_mostrar["L2"] = tabla_resultados_mostrar["L2"].map("{:.4f}".format)
for columna in ["MAE", "RMSE"]:
    tabla_resultados_mostrar[columna] = (
        tabla_resultados_mostrar[columna].map("{:,.0f}".format)
    )
tabla_resultados_mostrar["sMAPE (%)"] = (
    tabla_resultados_mostrar["sMAPE (%)"].map("{:.2f}".format)
)
tabla_resultados_mostrar["Segundos"] = (
    tabla_resultados_mostrar["Segundos"].map("{:.2f}".format)
)
display(tabla_resultados_mostrar)
```

# 9. Interpretación del tuneo interno

La tabla anterior permite responder, de manera provisional, cuál configuración aprende mejor cada serie:

- **El RMSE** penaliza con mayor fuerza los errores grandes y es el criterio de selección.
- **El MAE** expresa cuántos visitantes se desvía el modelo en un mes típico.
- **El sMAPE** permite comparar series de escalas distintas.
- **Las curvas de aprendizaje** permiten revisar si la mejora se debe a aprendizaje real o a sobreajuste.

La validación incluye marzo de 2020 a marzo de 2021, meses que no se parecen al historial usado para ajustar los pesos. Por eso las curvas muestran una pérdida de validación mucho mayor que la de ajuste y las predicciones no anticipan el cierre extraordinario. El diagnóstico por régimen cuantifica este efecto. # 10. Evaluación definitiva sobre el conjunto de prueba

Una vez elegida la configuración ganadora de cada serie, se ajusta un modelo nuevo con los **147 meses completos de entrenamiento**. El número de épocas se fija con la mejor época encontrada durante el tuneo interno. De esta manera, el test no participa en ninguna decisión.

La predicción de los 63 meses es recursiva:

1. se pronostica el primer mes usando únicamente el historial de entrenamiento;
2. ese pronóstico se agrega temporalmente al historial;
3. el procedimiento se repite hasta completar todo el horizonte.

Este enfoque reproduce el escenario real en el que no se conocen los valores futuros.


```python
def config_por_id(modelo_id):
    return next(c for c in CONFIGURACIONES if c["modelo"] == modelo_id)


def entrenar_y_pronosticar_test(serie, fila_ganadora):
    config = config_por_id(fila_ganadora["Modelo"])
    ventana = int(config["ventana"])
    train = serie.iloc[:N_TRAIN_ORIGINAL].astype(float)
    test = serie.iloc[N_TRAIN_ORIGINAL:].astype(float)

    scaler = MinMaxScaler(feature_range=(-1, 1))
    train_escalado = scaler.fit_transform(
        train.to_numpy().reshape(-1, 1)
    ).ravel()
    X_train = np.asarray([
        train_escalado[t - ventana:t]
        for t in range(ventana, len(train_escalado))
    ])
    y_train = train_escalado[ventana:]

    modelo = LSTMNumPyRegressor(
        hidden_sizes=config["hidden_sizes"],
        learning_rate=config["learning_rate"],
        l2=config["l2"],
        batch_size=config["batch_size"],
        max_epochs=int(fila_ganadora["Mejor época"]),
        patience=1,
        clip_norm=5.0,
        seed=SEED + sum(ord(c) for c in fila_ganadora["Serie"] + config["modelo"]),
    )
    modelo.fit_fixed_epochs(
        X_train, y_train, epochs=int(fila_ganadora["Mejor época"])
    )

    historial = list(train_escalado)
    pred_escalada = []
    for _ in range(len(test)):
        entrada = np.asarray([historial[-ventana:]], dtype=float)
        pred = float(modelo.predict(entrada)[0])
        pred_escalada.append(pred)
        historial.append(pred)

    prediccion = scaler.inverse_transform(
        np.asarray(pred_escalada).reshape(-1, 1)
    ).ravel()
    prediccion = np.clip(prediccion, 0, None)
    real = test.to_numpy()

    metricas = {
        "Serie": fila_ganadora["Serie"],
        "Modelo LSTM": fila_ganadora["Modelo"],
        "Ventana": ventana,
        "Épocas finales": int(fila_ganadora["Mejor época"]),
        "MAE": mean_absolute_error(real, prediccion),
        "RMSE": np.sqrt(mean_squared_error(real, prediccion)),
        "NRMSE (%)": 100 * np.sqrt(mean_squared_error(real, prediccion))
        / max(float(np.mean(np.abs(real))), 1e-9),
        "sMAPE (%)": smape(real, prediccion),
    }
    return metricas, pd.Series(prediccion, index=test.index), modelo


resultados_test_lstm = []
predicciones_test_lstm = {}
modelos_finales_lstm = {}
for _, fila in ganadores.iterrows():
    metricas, prediccion, modelo = entrenar_y_pronosticar_test(
        series[fila["Serie"]], fila
    )
    resultados_test_lstm.append(metricas)
    predicciones_test_lstm[fila["Serie"]] = prediccion
    modelos_finales_lstm[fila["Serie"]] = modelo

tabla_test_lstm = pd.DataFrame(resultados_test_lstm).sort_values("Serie")
tabla_test_lstm_mostrar = tabla_test_lstm.copy()
for columna in ["MAE", "RMSE"]:
    tabla_test_lstm_mostrar[columna] = (
        tabla_test_lstm_mostrar[columna].map("{:,.0f}".format)
    )
for columna in ["NRMSE (%)", "sMAPE (%)"]:
    tabla_test_lstm_mostrar[columna] = (
        tabla_test_lstm_mostrar[columna].map("{:.2f}".format)
    )
display(tabla_test_lstm_mostrar)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(15, 9), sharex=True)
for ax, (nombre, serie), color in zip(
    axes, series.items(), [PALETA["azul"], PALETA["turquesa"]]
):
    train = serie.iloc[:N_TRAIN_ORIGINAL]
    test = serie.iloc[N_TRAIN_ORIGINAL:]
    pred = predicciones_test_lstm[nombre]
    ax.plot(train.iloc[-36:], color=PALETA["gris"], lw=1.4,
            label="Últimos 36 meses de train")
    ax.plot(test, color=color, lw=2, label="Test observado")
    ax.plot(pred, color=PALETA["naranja"], lw=2.2, ls="--",
            label="Pronóstico LSTM")
    ax.axvline(test.index.min(), color=PALETA["rojo"], ls=":", lw=1.3)
    fila = tabla_test_lstm.loc[tabla_test_lstm["Serie"].eq(nombre)].iloc[0]
    ax.set_title(f"{nombre} · RMSE test={fila['RMSE']:,.0f}")
    ax.set_ylabel("Visitantes")
    ax.yaxis.set_major_formatter(lambda x, pos: f"{x/1000:.0f} mil")
    ax.legend(loc="upper left", ncol=3)
axes[-1].set_xlabel("Fecha")
plt.suptitle("Pronóstico recursivo definitivo de los 63 meses de prueba",
             fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

# 12. Análisis de similitud mediante catch22

## 12.1 Idea central

`catch22` es un conjunto canónico de 22 características que resume propiedades diversas de una serie: autocorrelación, periodicidad, distribución, regularidad, comportamiento de valores extremos, estructura simbólica y fluctuaciones. Su ventaja es convertir series completas en vectores numéricos comparables sin depender únicamente de una gráfica.

Se analizan las siete series construidas en el Laboratorio 1:

1. total nacional;
2. tres regiones geográficas con mayor acumulado;
3. tres fronteras con mayor acumulado.

Antes de extraer las características, cada serie se normaliza internamente con media cero y desviación uno. Después, como solicita el PDF, la matriz de características también se estandariza por columna antes de PCA, clustering y distancias.


```python
# pycatch22 implementa las 22 funciones canónicas en C.
ruta_vendor = Path.cwd() / "lab2_advance" / "vendor"
if ruta_vendor.exists():
    sys.path.insert(0, str(ruta_vendor))

try:
    from pycatch22 import catch22_all
    origen_catch22 = "pycatch22 disponible"
except ImportError:
    print("Instalando pycatch22 para reproducir las 22 características...")
    entorno = dict(**os.environ)
    entorno.setdefault("CC", "gcc")
    entorno.setdefault("PIP_CACHE_DIR", str(Path.cwd() / ".pip_cache"))
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "pycatch22==0.4.5"],
        env=entorno,
    )
    from pycatch22 import catch22_all
    origen_catch22 = "pycatch22 0.4.5 instalado"

print(origen_catch22)
```

```python
top_regiones = (
    visitantes.groupby("Región dos")["Viajero"].sum().nlargest(3).index
)
top_fronteras = (
    visitantes.groupby("Frontera")["Viajero"].sum().nlargest(3).index
)

series_catch22 = {"Total nacional": serie_total.copy()}
categoria_catch22 = {"Total nacional": "Total"}

for region in top_regiones:
    nombre = f"Región: {region}"
    series_catch22[nombre] = (
        visitantes.loc[visitantes["Región dos"].eq(region)]
        .groupby("Fecha")["Viajero"].sum()
        .reindex(calendario, fill_value=0)
        .astype(float)
    )
    categoria_catch22[nombre] = "Región"

for frontera in top_fronteras:
    nombre = f"Frontera: {frontera}"
    series_catch22[nombre] = (
        visitantes.loc[visitantes["Frontera"].eq(frontera)]
        .groupby("Fecha")["Viajero"].sum()
        .reindex(calendario, fill_value=0)
        .astype(float)
    )
    categoria_catch22[nombre] = "Frontera"

tabla_series_catch22 = pd.DataFrame({
    "Serie": list(series_catch22),
    "Categoría": [categoria_catch22[n] for n in series_catch22],
    "Meses": [len(s) for s in series_catch22.values()],
    "Inicio": [s.index.min().strftime("%Y-%m") for s in series_catch22.values()],
    "Fin": [s.index.max().strftime("%Y-%m") for s in series_catch22.values()],
})
display(tabla_series_catch22)
```

```python
def normalizar_serie_para_catch22(valores):
    valores = np.asarray(valores, dtype=float)
    desviacion = valores.std(ddof=0)
    if desviacion < 1e-12:
        return np.zeros_like(valores)
    return (valores - valores.mean()) / desviacion


filas_catch22 = {}
for nombre, serie in series_catch22.items():
    resultado = catch22_all(
        normalizar_serie_para_catch22(serie.values).tolist(),
        catch24=False,
    )
    filas_catch22[nombre] = dict(
        zip(resultado["names"], resultado["values"])
    )

matriz_catch22 = (
    pd.DataFrame.from_dict(filas_catch22, orient="index")
    .replace([np.inf, -np.inf], np.nan)
)
if matriz_catch22.isna().any().any():
    matriz_catch22 = matriz_catch22.fillna(matriz_catch22.median())

assert matriz_catch22.shape == (7, 22)
assert np.isfinite(matriz_catch22.to_numpy()).all()

print("Dimensión de la matriz catch22:", matriz_catch22.shape)
display(matriz_catch22.round(4))
```

# 15. LSTM enriquecido con características catch22

Para responder el inciso 2.14 se utiliza `01 La Aurora`, la serie más atípica según catch22. El modelo base ganador fue LSTM-2A.

El nuevo modelo conserva una ventana cruda de 12 meses, pero añade 22 características calculadas **únicamente con los 36 meses anteriores a cada objetivo**. Cada vector catch22 se repite a lo largo de los 12 pasos de la entrada, por lo que la LSTM recibe 23 variables por paso:

\[
\text{entrada}_t =
\left[
y_t,\,
f_{1,t}^{(catch22)},\ldots,
f_{22,t}^{(catch22)}
\right].
\]

Este diseño evita utilizar datos futuros. Las características se imputan y estandarizan usando exclusivamente el tramo disponible para entrenamiento.


```python
SERIE_ENRIQUECIDA = "01 La Aurora"
VENTANA_RAW = 12
CONTEXTO_CATCH22 = 36
CONFIG_ENRIQUECIDA = config_por_id("LSTM-2A")


def vector_catch22(valores):
    valores = normalizar_serie_para_catch22(valores)
    resultado = catch22_all(valores.tolist(), catch24=False)
    return np.asarray(resultado["values"], dtype=float)


def construir_entradas_enriquecidas(
    valores_escalados, posiciones, escalador_features=None,
    medianas=None, ajustar=False,
):
    X_raw = np.asarray([
        valores_escalados[t - VENTANA_RAW:t] for t in posiciones
    ])
    features = np.asarray([
        vector_catch22(
            valores_escalados[t - CONTEXTO_CATCH22:t]
        )
        for t in posiciones
    ])
    features[~np.isfinite(features)] = np.nan

    if ajustar:
        medianas = np.nanmedian(features, axis=0)
        medianas = np.where(np.isfinite(medianas), medianas, 0.0)
        features = np.where(np.isnan(features), medianas, features)
        escalador_features = StandardScaler().fit(features)
    else:
        features = np.where(np.isnan(features), medianas, features)

    features_z = escalador_features.transform(features)
    repetidas = np.repeat(
        features_z[:, None, :], VENTANA_RAW, axis=1
    )
    X_enriquecido = np.concatenate(
        [X_raw[:, :, None], repetidas], axis=2
    )
    return X_enriquecido, escalador_features, medianas
```

```python
serie_enriquecida = series[SERIE_ENRIQUECIDA]
train_original = serie_enriquecida.iloc[:N_TRAIN_ORIGINAL].astype(float)
scaler_raw_interno = MinMaxScaler(feature_range=(-1, 1))
scaler_raw_interno.fit(train_original.iloc[:N_CORE].values.reshape(-1, 1))
z_interno = scaler_raw_interno.transform(
    train_original.values.reshape(-1, 1)
).ravel()

posiciones_todas = np.arange(CONTEXTO_CATCH22, len(train_original))
pos_train = posiciones_todas[posiciones_todas < N_CORE]
pos_val = posiciones_todas[posiciones_todas >= N_CORE]

X_enr_train, scaler_feat_interno, medianas_internas = (
    construir_entradas_enriquecidas(
        z_interno, pos_train, ajustar=True
    )
)
X_enr_val, _, _ = construir_entradas_enriquecidas(
    z_interno, pos_val,
    escalador_features=scaler_feat_interno,
    medianas=medianas_internas,
    ajustar=False,
)
y_enr_train = z_interno[pos_train]
y_enr_val = z_interno[pos_val]

modelo_enriquecido_val = LSTMNumPyRegressor(
    hidden_sizes=CONFIG_ENRIQUECIDA["hidden_sizes"],
    learning_rate=CONFIG_ENRIQUECIDA["learning_rate"],
    l2=CONFIG_ENRIQUECIDA["l2"],
    batch_size=CONFIG_ENRIQUECIDA["batch_size"],
    max_epochs=90,
    patience=12,
    clip_norm=5.0,
    seed=SEED + 220,
)
modelo_enriquecido_val.fit(
    X_enr_train, y_enr_train, X_enr_val, y_enr_val
)

pred_enr_val_z = modelo_enriquecido_val.predict(X_enr_val)
pred_enr_val = scaler_raw_interno.inverse_transform(
    pred_enr_val_z.reshape(-1, 1)
).ravel()
real_enr_val = scaler_raw_interno.inverse_transform(
    y_enr_val.reshape(-1, 1)
).ravel()

comparacion_validacion_enriquecida = pd.DataFrame([
    {
        "Modelo": "LSTM-2A base",
        "MAE": ganadores.loc[
            ganadores["Serie"].eq(SERIE_ENRIQUECIDA), "MAE"
        ].iloc[0],
        "RMSE": ganadores.loc[
            ganadores["Serie"].eq(SERIE_ENRIQUECIDA), "RMSE"
        ].iloc[0],
    },
    {
        "Modelo": "LSTM-2A + catch22",
        "MAE": mean_absolute_error(real_enr_val, pred_enr_val),
        "RMSE": np.sqrt(mean_squared_error(real_enr_val, pred_enr_val)),
    },
])
display(comparacion_validacion_enriquecida.round(2))
print("Mejor época del modelo enriquecido:",
      modelo_enriquecido_val.best_epoch_)
print("Forma de entrada enriquecida:", X_enr_train.shape)
```

```python
# Reentrenamiento final con los 147 meses completos
train_enr = serie_enriquecida.iloc[:N_TRAIN_ORIGINAL].astype(float)
test_enr = serie_enriquecida.iloc[N_TRAIN_ORIGINAL:].astype(float)
scaler_raw_final = MinMaxScaler(feature_range=(-1, 1))
z_final = scaler_raw_final.fit_transform(
    train_enr.values.reshape(-1, 1)
).ravel()

pos_finales = np.arange(CONTEXTO_CATCH22, len(train_enr))
X_enr_final, scaler_feat_final, medianas_finales = (
    construir_entradas_enriquecidas(
        z_final, pos_finales, ajustar=True
    )
)
y_enr_final = z_final[pos_finales]

modelo_enriquecido_final = LSTMNumPyRegressor(
    hidden_sizes=CONFIG_ENRIQUECIDA["hidden_sizes"],
    learning_rate=CONFIG_ENRIQUECIDA["learning_rate"],
    l2=CONFIG_ENRIQUECIDA["l2"],
    batch_size=CONFIG_ENRIQUECIDA["batch_size"],
    max_epochs=modelo_enriquecido_val.best_epoch_,
    patience=1,
    clip_norm=5.0,
    seed=SEED + 220,
)
modelo_enriquecido_final.fit_fixed_epochs(
    X_enr_final, y_enr_final,
    epochs=modelo_enriquecido_val.best_epoch_,
)

historial_enr = list(z_final)
pred_enr_test_z = []
for _ in range(len(test_enr)):
    posicion = len(historial_enr)
    valores_historia = np.asarray(historial_enr, dtype=float)
    X_paso, _, _ = construir_entradas_enriquecidas(
        valores_historia,
        np.asarray([posicion]),
        escalador_features=scaler_feat_final,
        medianas=medianas_finales,
        ajustar=False,
    )
    pred_paso = float(modelo_enriquecido_final.predict(X_paso)[0])
    pred_enr_test_z.append(pred_paso)
    historial_enr.append(pred_paso)

pred_enr_test = scaler_raw_final.inverse_transform(
    np.asarray(pred_enr_test_z).reshape(-1, 1)
).ravel()
pred_enr_test = np.clip(pred_enr_test, 0, None)

metricas_enriquecido = {
    "Modelo": "LSTM-2A + catch22",
    "MAE": mean_absolute_error(test_enr.values, pred_enr_test),
    "RMSE": np.sqrt(mean_squared_error(test_enr.values, pred_enr_test)),
    "sMAPE (%)": smape(test_enr.values, pred_enr_test),
}

fila_base_test = tabla_test_lstm.loc[
    tabla_test_lstm["Serie"].eq(SERIE_ENRIQUECIDA)
].iloc[0]
fila_ses = referencia_lab1.loc[
    referencia_lab1["Serie"].eq(SERIE_ENRIQUECIDA)
].iloc[0]
comparacion_catch22_lstm = pd.DataFrame([
    {
        "Modelo": fila_base_test["Modelo LSTM"] + " base",
        "MAE": fila_base_test["MAE"],
        "RMSE": fila_base_test["RMSE"],
        "sMAPE (%)": fila_base_test["sMAPE (%)"],
    },
    metricas_enriquecido,
    {
        "Modelo": fila_ses["Modelo tradicional"],
        "MAE": fila_ses["MAE tradicional"],
        "RMSE": fila_ses["RMSE tradicional"],
        "sMAPE (%)": fila_ses["sMAPE tradicional (%)"],
    },
]).sort_values("RMSE")
display(comparacion_catch22_lstm.round(2))
```

## 15.1 Discusión del modelo enriquecido

El modelo con catch22 **no supera** a la LSTM base. Aunque las características resumen periodicidad, atípicos y dependencia, repetir el mismo vector a lo largo de cada secuencia aumenta la dimensión de entrada de 1 a 23 variables con una cantidad reducida de ejemplos. Esto puede introducir redundancia y facilitar el sobreajuste.

El resultado es útil: demuestra que una representación descriptiva eficaz para comparar series no necesariamente mejora un pronóstico recursivo. La LSTM base conserva una señal más directa y parsimoniosa. Para trabajos futuros podría utilizarse una arquitectura con dos ramas —una LSTM para la secuencia y una red densa para catch22— o ampliar el número de series y ventanas de entrenamiento.
